# Snowshoe hares at onanza Creek Experimental Forest
## Description
This document analyses data regarding snowshoe hares (_Lepus_ _americanus_) in the Bonanza Creek Experimental Forest in Alaska, USA.  Data was accessed from the Environmental Data Initiative and was created by Knut Kielland, Stuart Chapin, and Roger Ruess with the Bonanza Creek LTER to investigate snowshoe hare population densities and apparent decline.  Each row in the dataset cooresponds to one hare trapped in the Bonanza Creek National Forest between 06/01/1999 and 09/14/2002.  The dataset does not include sensitive data. The researchers used a caoture, recapture methodology to determine population density and recorded physical characteristics of the hares at time of capture. Hare wieght is recorded in grams.  Missing values are coded as NaN.  Methodologies are described in Flora (2002).
