### Data Source
The following analysis uses compiled occurance records for arthropod prey for engangered vertebrate species in California dylands, along with associated environmental data.  The dataset can be found in the [Knowledge Network for Biocomplexity](https://knb.ecoinformatics.org/view/doi%3A10.5063%2FF1VM49RH) repository from King et al. (2023).  

### Date of access: 
October 1, 2026

Rachel King, Jenna Braun, Michael Westphal, & CJ Lortie. (2023). Compiled occurrence records for prey items of listed species found in California drylands with associated environmental data.
